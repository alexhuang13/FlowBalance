```python
import json
import numpy as np

print(df['prompt'].head(1))
print(df['reward_model'].head(1))

# Convert any NumPy arrays to lists
def convert(obj):
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    raise TypeError

prompt = df.at[0, 'prompt']
json_str = json.dumps(prompt, indent=2, default=convert)
reward = df.at[0, 'reward_model']
print(json.dumps(reward, indent=2, default=convert))

import textwrap

print("\n".join(textwrap.wrap(json_str, width=120)))  # wrap at 120 chars
```
